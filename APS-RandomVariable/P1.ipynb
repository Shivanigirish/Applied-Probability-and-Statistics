```python
# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib for plotting graphs
import matplotlib.pyplot as plt

# Probability that a customer clicks the promotional link
p = 0.30

# Number of customers
sample_size = 20

print("Click probability:", p)
print("Number of customers:", sample_size)
```

```python
# Generate Bernoulli outcomes for 20 customers
# n=1 means one trial for each customer
# p=0.30 means 30% probability of clicking
# size=20 means generate results for 20 customers

samples = np.random.binomial(n=1, p=p, size=sample_size)

print("Generated observations:")
print(samples)
```

```python
# Count the number of customers who clicked
# In our data, click is represented by 1
number_of_clicks = np.sum(samples == 1)

# Count the number of customers who did not click
# In our data, non-click is represented by 0
number_of_non_clicks = np.sum(samples == 0)

print("Number of clicks:", number_of_clicks)
print("Number of non-clicks:", number_of_non_clicks)

# Check whether the total is 20
print("Total customers:", number_of_clicks + number_of_non_clicks)
```

```python
# Empirical PMF means probability calculated from our generated sample

# Probability of no click
empirical_pmf_0 = number_of_non_clicks / sample_size

# Probability of click
empirical_pmf_1 = number_of_clicks / sample_size

print("Empirical PMF:")
print("P(X=0) =", empirical_pmf_0)
print("P(X=1) =", empirical_pmf_1)
```

```python
# Theoretical probability of no click
theoretical_pmf_0 = 1 - p

# Theoretical probability of click
theoretical_pmf_1 = p

print("Theoretical PMF:")
print("P(X=0) =", theoretical_pmf_0)
print("P(X=1) =", theoretical_pmf_1)
```

```python
# Compare the empirical and theoretical probabilities

print("Comparison of PMFs")


print("P(X=0):")
print("Theoretical =", theoretical_pmf_0)
print("Empirical   =", empirical_pmf_0)

print("\nP(X=1):")
print("Theoretical =", theoretical_pmf_1)
print("Empirical   =", empirical_pmf_1)

# Calculate absolute differences
difference_0 = abs(theoretical_pmf_0 - empirical_pmf_0)
difference_1 = abs(theoretical_pmf_1 - empirical_pmf_1)

print("\nAbsolute difference:")
print("For X=0:", difference_0)
print("For X=1:", difference_1)
```

```python
# Values of X
x = [0, 1]

# Empirical probabilities
empirical_values = [empirical_pmf_0, empirical_pmf_1]

# Theoretical probabilities
theoretical_values = [theoretical_pmf_0, theoretical_pmf_1]

# Width of each bar
width = 0.35

# Plot empirical PMF
plt.bar(
    np.array(x) - width/2,
    empirical_values,
    width,
    label="Empirical"
)

# Plot theoretical PMF
plt.bar(
    np.array(x) + width/2,
    theoretical_values,
    width,
    label="Theoretical"
)

# Add labels and title
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Empirical vs Theoretical PMF")

# Rename 0 and 1
plt.xticks([0, 1], ["No Click (0)", "Click (1)"])

# Show legend
plt.legend()

# Display graph
plt.show()
```

```python
# Theoretical expectation of a Bernoulli random variable
# E(X) = p

theoretical_mean = p

# Empirical expectation is the mean of generated observations
empirical_mean = np.mean(samples)

print("Theoretical expectation:", theoretical_mean)
print("Empirical expectation:", empirical_mean)
```

```python
# Theoretical variance of Bernoulli distribution
# Var(X) = p(1-p)

theoretical_variance = p * (1 - p)

# Empirical variance calculated from generated observations
empirical_variance = np.var(samples)

print("Theoretical variance:", theoretical_variance)
print("Empirical variance:", empirical_variance)
```

```python
# Display all important results together

print("Final Result")
print("=" * 70)

print("Measure          Theoretical    Empirical    Absolute Difference")
print("-" * 70)

print(
    "P(X=0)           ",
    theoretical_pmf_0,
    "       ",
    empirical_pmf_0,
    "       ",
    abs(theoretical_pmf_0 - empirical_pmf_0)
)

print(
    "P(X=1)           ",
    theoretical_pmf_1,
    "       ",
    empirical_pmf_1,
    "       ",
    abs(theoretical_pmf_1 - empirical_pmf_1)
)

print(
    "Expectation       ",
    theoretical_mean,
    "       ",
    empirical_mean,
    "       ",
    abs(theoretical_mean - empirical_mean)
)

print(
    "Variance          ",
    theoretical_variance,
    "       ",
    empirical_variance,
    "       ",
    abs(theoretical_variance - empirical_variance)
)
```

```python
# Check how the results change when we increase the sample size

sample_sizes = [10, 100, 1000, 10000]

print("Effect of Sample Size")
print("=" * 55)
print("Sample Size    Empirical Mean    Empirical Variance")
print("-" * 55)

# Repeat the experiment for different sample sizes
for n in sample_sizes:

    # Generate Bernoulli observations
    samples_n = np.random.binomial(
        n=1,
        p=p,
        size=n
    )

    # Calculate empirical mean
    mean_n = np.mean(samples_n)

    # Calculate empirical variance
    variance_n = np.var(samples_n)

    print(
        n,
        "          ",
        round(mean_n, 4),
        "          ",
        round(variance_n, 4)
    )
```
