```python
# Problem Set 2
## ML Use Case: Number of Fraudulent Transactions Flagged
```

```python
# Number of transactions examined every hour
n_transactions = 50

# Probability that a transaction is flagged
p_flagged = 0.04

# Number of one-hour periods to simulate
n_hours = 20000

print("Transactions per hour:", n_transactions)
print("Flagging probability:", p_flagged)
print("Number of simulated hours:", n_hours)
```

```python
# Simulate the number of flagged transactions for 20,000 hours
# Binomial distribution is used because:
# - There are 50 transactions in every hour
# - Each transaction has probability 0.04 of being flagged
import numpy as np
flagged_transactions = np.random.binomial(
    n=n_transactions,
    p=p_flagged,
    size=n_hours
)

print("First 20 simulated hours:")
print(flagged_transactions[:20])
```

```python
# Calculate empirical mean
empirical_mean = np.mean(flagged_transactions)

# Calculate empirical variance
empirical_variance = np.var(flagged_transactions)

# Calculate theoretical mean
theoretical_mean = n_transactions * p_flagged

# Calculate theoretical variance
theoretical_variance = (
    n_transactions
    * p_flagged
    * (1 - p_flagged)
)

print("Theoretical Mean:", theoretical_mean)
print("Empirical Mean:", empirical_mean)

print("\nTheoretical Variance:", theoretical_variance)
print("Empirical Variance:", empirical_variance)
```

```python
# The investigation team can examine at most 4 flagged transactions per hour

# Count hours where more than 4 transactions were flagged
hours_more_than_4 = np.sum(flagged_transactions > 4)

# Calculate the proportion of such hours
prob_more_than_4 = hours_more_than_4 / n_hours

print("Hours with more than 4 flagged transactions:", hours_more_than_4)

print(
    "Empirical probability of more than 4 flagged transactions:",
    prob_more_than_4
)
```

```python
# Plot the distribution of flagged transactions per hour
import matplotlib.pyplot as plt

plt.hist(
    flagged_transactions,
    bins=np.arange(flagged_transactions.max() + 2) - 0.5,
    edgecolor="black"
)

plt.xlabel("Number of Flagged Transactions")
plt.ylabel("Number of Hours")
plt.title("Distribution of Flagged Transactions per Hour")

plt.show()
```
