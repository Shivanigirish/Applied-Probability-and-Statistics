```python
import random
n=100
heads = 0
tails = 0

for i in range(n):
    toss = random.choice(["h", "t"])

    if toss == 'h':
        heads +=1
    else:
        tails +=1

print("number of heads",heads)
print("number of tails",tails)

exp_prob_h = heads/n
exp_prob_t = tails/n

print("expermental probability of head",exp_prob_h)
print("experimental probability of tail",exp_prob_t)
```

```python
import random
for n in [10, 100, 1000, 10000]:
    heads = 0
    tails = 0

    for i in range(n):
        toss = random.choice(["h", "t"])

        if toss =='h':
            heads += 1
        else:
            tails += 1

    print("trials=", n)
    print("p(head) =",heads/n)
    print("p(tail) =",tails/n)
    print()
```

```python
import random

n = 10000
a = 0
b = 0
ab = 0

for i in range(n):
    toss = random.randint(1,6)

    if toss in [1,2,3]:
        a = a+1
    if toss in [5,6]:
        b=b+1
    if toss in [1,2,3,5,6]:
        ab=ab+1

pa=a/n
pb=b/n 
pab=ab/n 

print("p(a)=",pa)
print("p(b)=",pb)
print("p(a union b)=",pab)
print("p(a)+p(b)=",pa+pb)
```
